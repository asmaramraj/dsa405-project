This dataset includes defensive data for every regular season game played by every NBA team in the 2025-2026 season on points in the paint against an opposing team, as well as on neutral/away/away sites. This data was collected by the NBA statisticians and placed on Kaggle as a public sports analysis resource because there are restrictions on scraping data off of NBA.com.

Each row represents a regular season game for each team in the NBA and includes data on opposing team scoring and on the site of the game. This data can be used effectively to analyze overall defensive performance at home versus away but is not suitable for analyzing play by play details such as a specific defender on a specific player, fouls and foul problems, or specific defensive calls made on a specific possession.

This dataset can be used to effectively address the main objective of this project: did the Charlotte Hornets allow a significantly higher number of points in the paint on an away site compared to a home site during the 2025-2026 NBA season?
